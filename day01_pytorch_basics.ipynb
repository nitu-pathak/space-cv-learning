Tensor Basics

```python
import torch

# Create tensors
x = torch.tensor([[1.0, 2.0], [3.0, 4.0]])
print(x.shape)       # What does this tell you?
print(x.dtype)
print(x.device)

# Simulate a tiny satellite image: 3 bands, 4x4 pixels
satellite_image = torch.randn(3, 4, 4)
print(satellite_image.shape)  # [Channels, Height, Width]
print(satellite_image)
```

GPU Check

```python
# Check if GPU is available on Colab
print(torch.cuda.is_available())

# Go to Runtime > Change Runtime Type > T4 GPU, then rerun
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

satellite_image = satellite_image.to(device)
print(satellite_image.device)
```

Autograd

```python
# requires_grad tells PyTorch to track this tensor
x = torch.tensor(2.0, requires_grad=True)

# Simple operation
y = x ** 2 + 3 * x + 1   # y = x² + 3x + 1

# Compute gradient dy/dx
y.backward()
print(x.grad)   # Should be 2*2 + 3 = 7.0  ← verify this manually
```

Space Framing

```python
# Simulate a 13-band Sentinel-2 image patch
sentinel2_patch = torch.randn(13, 64, 64)  # 13 bands, 64x64 pixels
print(f"Sentinel-2 patch shape: {sentinel2_patch.shape}")
print(f"Total values in this patch: {sentinel2_patch.numel()}")
print(f"Memory (bytes): {sentinel2_patch.element_size() * sentinel2_patch.numel()}")
```
